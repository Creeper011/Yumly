# Yumly Playground

This notebook provides a hands-on introduction to the `yumly` configuration language. We will explore how to use the `yumly` library to parse structured string data directly within Python, demonstrating its `loads` method for in-memory configuration parsing.

### First, in a case if you have only the content, not the file:

```python
from yumly import Yumly

# Define a string that contains key-value pairs with type annotations.
data = "projectName ;string = \"Yumly\", version ;string = \"0.0.1\""
yumly = Yumly()

# Use the 'loads' method to parse the 'data' string into a Python dictionary.
print(yumly.loads(data)) # This returns an dict[str, Any]
```

*also, we can validate this content:*

```python
from yumly import Yumly

data = "projectName ;string = \"Yumly\", version ;string = \"0.0.1\""
yumly = Yumly()

# Use the `validate_content` method to validate
print(yumly.validate_content(data)) # This returns an bool
```

### Now for parse an yumly file, we need to create an `yumly` file.

```python
FILENAME = "config.yumly"

data = """
projectName ;string = "Yumly",
version ;string = "0.0.1"

(randomBlock) {
  randomKey ;string = "randomValue",
  randomList ;list[string] = ["item1", "item2", "item3"]
}
"""

with open(FILENAME, 'w') as f:
    f.write(data)

print(f'File {FILENAME} created successfully.')
```

*Now, the parsing of it:*

```python
from yumly import Yumly

yumly = Yumly()

print(yumly.load("config.yumly")) # this returns an dict[str, Any]
```

*also, we can validate the file*

```python
from yumly import Yumly

yumly = Yumly()

print(yumly.validate_file("config.yumly")) # this returns an bool
```

### And we can write yumly data, using `dump` or `dumps` methods

```python
from yumly import Yumly

# structured data
data = {
    "projectName": "Yumly",
    "version": "0.0.1",
    "randomBlock": {
        "randomKey": "randomValue",
        "randomList": ["item1", "item2", "item3"]
    }
}

yumly = Yumly()

print("Dumped data:")
print("-"*20)
print(yumly.dumps(data)) # this returns an str

with open("dumpedYumly.yumly", 'w') as f:
    yumly.dump(data, f) # actually this read the entire file

print("")
print(f"Content of dumpedYumly.yumly:")
print("-"*20)
with open("dumpedYumly.yumly", 'r') as f:
    print(f.read())
    f.close()

```
